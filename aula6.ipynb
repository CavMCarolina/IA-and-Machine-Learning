```python
import matplotlib.pyplot as plt
```

```python
lista_x = [5, 8, 11, 9, 4, -3, 2, 17, 6, 7]
lista_y = [3, -1, 6, 5, 7, 2, 14, 9, 8, 1]

def pitagoras(lista_x, lista_y):
    matriz = []

    # Para cada ponto fixado, calcular as distâncias com todos os outros pontos e guardar em uma linha
    for p in range(len(lista_x)):
        linha = []

        for x in range(len(lista_x)):
            distancia = ((lista_x[p] - lista_x[x])**2 + (lista_y[p] - lista_y[x])**2)**0.5
            distancia = round(distancia, 2)
            linha.append(distancia)
        
        # Adicionar linha para cada ponto --> Matriz com diagonal 0 simétrica
        matriz.append(linha)

    plt.plot(lista_x, lista_y, 'bo')
    plt.show()

    return matriz



pitagoras(lista_x, lista_y)
```
