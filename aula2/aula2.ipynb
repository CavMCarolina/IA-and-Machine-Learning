# Meu Código

```python
data = pd.read_excel("./data.xlsx")

data_x = data['x']
data_y = data['y']
```

```python
def get_a_b(listaX, listaY):
    mediaX = sum(listaX) / len(listaX)
    mediaY = sum(listaY) /len(listaY)

    listaSoma = []
    listaMultX = []
    listaMultY = []
    listaDiv = []
    for x in listaX:
        param1 = x - mediaX
        listaMultX.append(param1)

        div = (x - mediaX) ** 2

        listaDiv.append(div)

    for y in listaY:
        param2 = y - mediaY
        listaMultY.append(param2)

        
    for n in range(len(listaMultX)):
        mult = listaMultX[n] * listaMultY[n]

        listaSoma.append(mult)

    a = sum(listaSoma) / sum(listaDiv)

    b = mediaY - a * mediaX
    
    return a, b
```

```python
a, b = get_a_b(data_x, data_y)
print(f'~~~ Resultado: ~~~\nA = {a}\nB = {b}')
```

```python
def calculate(x, y):
    x = np.array(x)
    y = np.array(y)

    x_bar = np.mean(x)
    y_bar = np.mean(y)

    a = np.dot(x - x_bar, y - y_bar) / np.dot(x - x_bar, x - x_bar)
    print(a)

    b = y_bar - a * x_bar
    print(b)

    plt.plot(x, a * x + b)
    plt.plot(x, y)
    plt.show()

calculate(data_x, data_y)
```
